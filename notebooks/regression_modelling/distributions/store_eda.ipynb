## 2. National distribution — counts per brand and per state


```python
counts = eda.store_counts(STORE)
print(f"total stores: {counts['total_stores'].iloc[0]:,}" if len(counts) else "no rows")
counts.head(20)
```

```python
states = eda.state_counts(STORE)
states.groupby('state_code')['cnts'].sum().sort_values(ascending=False).head(15)
```
